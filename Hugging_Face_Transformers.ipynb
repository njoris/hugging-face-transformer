```python
import transformers
from transformers import pipeline
print(transformers.__version__)
```

```python
classifier = pipeline("sentiment-analysis")
classifier(
    "Tout va bien aujourd'hui !"
)
```

```python
classifier = pipeline('zero-shot-classification')
classifier(
            "Hésite pas a faire appel a ta banque pour ouvrir un PEA",

            candidate_labels=['economy','politic' ,'education']
)
```

```python
generator = pipeline('text-generation')
generator(
    "Aujourd'hui je vais developper une ",

    max_length=20,
    num_return_sequences=2,
)

```

```python
unmasker = pipeline("fill-mask" , model="bert-base-cased")
unmasker("This course will teach you all about [MASK] models.", top_k=2)
```

```python
ner = pipeline("ner", aggregation_strategy="simple")
ner(
    "Je m'appel Joris je travail a Inelys sur Lyon"
)
```

```python
question_answerer = pipeline("question-answering")
question_answerer(
    question="Where do I work?",
    context="My name is Joris and I work at Inelys in Lyon"
)
```
